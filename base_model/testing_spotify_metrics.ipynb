```python
import matplotlib.pyplot as plt
import pandas as pd

def plot_multiple_artists(artist_data, y_col="pop_mean", title="Artist popularity by career year"):
    """
    artist_data: dict mapping artist name -> list of yearly records
        e.g. {
          "Artist A": [{'year': 1982, 'pop_mean': 34.8, ...}, ...],
          "Artist B": [{'year': 1990, 'pop_mean': 42.1, ...}, ...],
        }
    y_col: metric to plot on y-axis
    """
    plt.figure(figsize=(9, 6))

    for artist, records in artist_data.items():
        df = pd.DataFrame(records).dropna(subset=["year", y_col]).sort_values("year").reset_index(drop=True)
        if df.empty:
            continue

        # Compute "career year index" — increments only for years with data
        df["career_year"] = range(1, len(df) + 1)

        plt.scatter(df["career_year"], df[y_col], label=artist, alpha=0.8)

    plt.title(title)
    plt.xlabel("Career year (debut = 1)")
    plt.ylabel(y_col.replace("_", " ").title())
    plt.legend(title="Artist")
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
```
