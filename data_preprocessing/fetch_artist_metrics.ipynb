```python
with open('data/artist_to_metric_sources_map.json') as f:
    data = json.load(f)
```

```python
import re

# Spotify uses Base62: 0-9, A-Z, a-z
BASE62_PATTERN = re.compile(r'^[0-9A-Za-z]{22}$')

def is_spotify_artist_id(s: str) -> bool:
    """
    Checks if a given string is a valid Spotify artist ID.
    Spotify IDs are 22-character Base62-encoded values.

    Args:
        s: The string to check.

    Returns:
        True if the string matches Spotify's ID format, False otherwise.
    """
    return bool(BASE62_PATTERN.match(s))
```

```python
is_spotify_artist_id('0wlw3iBVFnAzSviZyLjxDJ')
```

```python
mbids_with_valid_spotify_id = []

for mbid, sources in data.items():
    if 'spotify' not in sources:
        continue
    source = sources['spotify']
    spotify_id = source.split('/')[-1]
    if len(spotify_id) != 22 or not is_spotify_artist_id(spotify_id):
        continue
    mbids_with_valid_spotify_id.append((mbid, spotify_id))

len(mbids_with_valid_spotify_id)
```

```python
import pandas as pd
```

```python
pd.DataFrame(mbids_with_valid_spotify_id).rename(columns={0:'mbid', 1:'spotify_id'}).to_csv('data/all_spotify_filtered_artist_ids.csv', index=False)
```
