```python
API_BASE = "https://ws.audioscrobbler.com/2.0/"
import requests
api_key = os.getenv('LASTFM_API_KEY')
def fetch_lastfm_stats(mbid: str, timeout: float = 10.0):
    """
    Returns (listeners, playcount) for an artist MBID.
    On any error, returns (None, None).
    """
    try:
        r = requests.get(
            API_BASE,
            params={
                "method": "artist.getInfo",
                "api_key": api_key,
                "mbid": mbid,
                "format": "json",
            },
            timeout=timeout,
        )
        r.raise_for_status()
        data = r.json()
        if "error" in data:
            # e.g., unknown artist/invalid MBID; keep graceful
            return (None, None)
        artist = data.get("artist", {}) or {}
        stats = artist.get("stats", {}) or {}
        listeners = int(stats.get("listeners", 0)) if stats.get("listeners") is not None else None
        playcount = int(stats.get("playcount", 0)) if stats.get("playcount") is not None else None
        return (listeners, playcount)
    except Exception:
        return (None, None)
```
