```python
first_n_lines_data = []
with open(output_file, 'r', encoding='utf-8') as f:
    for i, line in enumerate(f):
        if i >= 2000:
            break
        try:
            first_n_lines_data.append(json.loads(line.strip()))
        except json.JSONDecodeError as e:
            print(f"Error decoding JSON on line {i+1}: {line.strip()} - {e}")
```
