```python

def normalize_uuid(u):
    return str(uuid.UUID(u))

# Example usage:
raw = "550e8400e29b41d4a716446655440000"
print(normalize_uuid(raw)) 
```

```python
with open('data/artist_mbids.txt') as f:
    artist_mbids = f.read().splitlines()
```

```python
for mbid in artist_mbids:
    if normalize_uuid(mbid) != mbid:
        print(mbid)
```
