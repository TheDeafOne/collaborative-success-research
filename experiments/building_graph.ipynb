```python
family_data = []
bad_lines = []

with open('../data/artist_collab_data/filtered_artists_with_wikidata.jsonl', 'r') as f:
    for i, line in enumerate(f, start=1):
        line = line.strip()
        if not line:
            continue  # skip blank lines
        try:
            obj = json.loads(line)
            family_data.append(obj)
        except json.JSONDecodeError as e:
            print(f"Skipping bad line {i}: {e}")
            bad_lines.append((i, line))
```

```python
import networkx as nx
import matplotlib.pyplot as plt

# Create the graph
G = nx.Graph()

# Add edges where has_mbid is True
for person in family_data:
    source_mbid = person["mbid"]
    for rel in person["relatives"]:
        if rel["has_mbid"]:
            G.add_edge(source_mbid, rel["relative_mbid"], relation=rel["relation"], label=rel["label"])

```

```python
# Draw the graph
plt.figure(figsize=(8, 6))
pos = nx.spring_layout(G, seed=42)
nx.draw(G, pos, with_labels=False, node_size=1, node_color="skyblue", edge_color="gray")

# Add relation labels
edge_labels = nx.get_edge_attributes(G, "relation")
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_color="darkred")

plt.title("Network of Individuals (has_mbid=True)")
plt.show()
```
