```python
import pandas as pd
```

```python
base_features = pd.read_csv('../data/artist_base_features_5_year.csv')
network_features = pd.read_csv('../data/graphs/better_graph/node_features.csv')
```

```python
merged = base_features.merge(network_features, how='left', left_on='artist_mbid', right_on='node')
```

```python
merged = merged[merged['closeness'].notna()]
```

```python
merged.to_csv('../data/all_features_5_year.csv')
```

```python
filtered_artist_mbids = set(merged['artist_mbid'])
```

```python
base_features = base_features[base_features['artist_mbid'].isin(filtered_artist_mbids)]
```
